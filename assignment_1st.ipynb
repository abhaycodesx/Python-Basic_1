```python
print("""Q6. Write a function to demonstrate the behavior of mutable and immutable arguments?""")

print("""function for mutable arguments""")


print("Function for mutable arguments : ")
def mutable_arguments(arguments):
    print(f"This is our given arguments {arguments}")
    arguments_before = id(arguments)
    print(f"This is the id of {arguments} before modifying {arguments_before}")
    
    arguments[1] += 10
    print(f"This is our arguments after modifying {arguments}")
    arguments_after = id(arguments)
    print(f"This is the id of {arguments} after modification {arguments_after}")
    
    """here we can see that id of arguments_before is same as the id of arguments_after means mutable argumetns are still pointing towards same memory location"""
    if arguments_before == arguments_after:
        print("id is same")
    
list_argument = [1,2,3,4]
mutable_arguments(list_argument)

print()
print()

print("Function for immutable arguments : ")

def immutable_arguments(argument):
    print(f"This is our given arguments {argument}")
    arguments_before = id(argument)
    print(f"This is the id of {argument} before modification {arguments_before}")
    
    argument += 10
    print(f"This is our argument after modification {argument}")
    arguments_after = id(argument)
    print(f"This is the id of arguments after modification {arguments_after}")
    
    if arguments_before != arguments_after:
        print("Id is different")
    
variable_immutable = 10

immutable_arguments(variable_immutable)
    
```

```python
print("""Q14. Write a program to reverse a string using a while loop?""")

name_str = input("Enter the string : ")
reverse_str = ""
num = len(name_str)

while num-1>=0:
    reverse_str += name_str[num-1]
    num -= 1

print(name_str)
print(reverse_str)
```

```python
print("""Q15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.""")
num = int(input("Enter the number : "))
fact = 1

while num>0:
    fact = fact*num
    num -=1
print(f"The factorial of given number {num} is {fact}")
```
